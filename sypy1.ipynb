```python
number = int(input("Enter a number: "))
sum_digits = 0
while number > 0:
    digit = number % 10
    sum_digits = sum_digits + digit
    number = number // 10
    print("Sum of digits = ", sum_digits)
```

```python
import numpy as np

# Creating array
a = np.array([10, 20, 30, 40, 50])
print("Array: ", a)

# Basic Operations
print("Addition: ", a + 5)
print("Multiplication: ",a * 2)

# Indexing
print("First Element: ",a[0])
print("Last Element: ",a[-1])

# Slicing
print("Elements from index 1 to 3: ",a[1:4])
print("Alternate Elements: ",a[::2])
```

```python
number = int(input("Enter a number: "))
if number % 2 == 0:
    print("The number is Even")

else:
    print("The number is Odd")
```

```python
n = int(input("Enter number of terms: "))
a = 0
b = 1
print("Fibonacci Series: ")
for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
sgpi = float(input("Enter SGPI: "))
if sgpi >= 9.00 and sgpi <= 10.00:
    print("Grade: O")
elif sgpi >= 8.00 and sgpi <= 8.99:
    print("Grade: A+")
elif sgpi >= 7.00 and sgpi <= 7.99:
    print("Grade: A")
elif sgpi >= 6.00 and sgpi <= 6.99:
    print("Grade: B+")
elif sgpi >= 5.50 and sgpi <= 5.99:
    print("Grade: B")
elif sgpi >= 5.00 and sgpi <= 5.49:
    print("Grade: C")
elif sgpi >= 4.00 and sgpi <= 4.99:
    print("Grade: P")
else:
    print("Grade: F")
```

```python
from datetime import datetime

name = input("Enter your name: ")

age = int(input("Enter your age: "))

current_year = datetime.now().year
year_100 = current_year + (100 - age)

print(name,"will turn 100 years old in",year_100)
```
